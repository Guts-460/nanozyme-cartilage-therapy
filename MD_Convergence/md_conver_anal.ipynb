```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import os

# 设置中文字体支持（如果需要）
# matplotlib.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
matplotlib.rcParams['axes.unicode_minus'] = False

# 定义三种模式
modes = ['superficial', 'middle', 'deep']
# 定义对应的颜色和标签
colors = {
    'superficial': "#03E2F2",
    'middle': "#C9C509",
    'deep': "#F37A00"
}
labels = {
    'superficial': 'Superficial',
    'middle': 'Middle',
    'deep': 'Deep'
}

def read_rg_data(file_path):
    """读取b-rg.xvg文件，返回时间和Rg值"""
    time_ps = []
    rg_values = []
    
    with open(file_path, 'r') as f:
        for line in f:
            # 跳过注释行（以#或@开头的行）
            if line.startswith('#') or line.startswith('@'):
                continue
            # 分割数据行
            data = line.strip().split()
            if len(data) >= 2:  # 确保至少有2列数据
                time_ps.append(float(data[0]))  # 第一列是时间(ps)
                rg_values.append(float(data[1]))  # 第二列是Rg值
    
    return np.array(time_ps), np.array(rg_values)

# 创建1×3的子图
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 为每个模式单独绘图
for idx, mode in enumerate(modes):
    ax = axes[idx]
    file_path = f'GAGs/b-{mode}-rg.xvg'
    
    try:
        time_ps, rg_values = read_rg_data(file_path)
        time_ns = time_ps / 1000.0  # 将时间从皮秒转换为纳秒
        
        ax.plot(time_ns, rg_values, 
                color=colors[mode], 
                linestyle='-', 
                linewidth=1.5, 
                label=labels[mode])
        
        print(f"成功读取 {mode} 数据，共 {len(rg_values)} 个数据点")
        
    except FileNotFoundError:
        print(f"警告：文件 {file_path} 未找到，跳过")
    except Exception as e:
        print(f"读取 {file_path} 时出错：{e}")
    
    # 设置每个子图的坐标轴标签
    ax.set_xlabel('Time (ns)', fontsize=28)
    ax.set_ylabel('Rg (nm)', fontsize=28)
    
    # 设置坐标轴范围
    ax.set_xlim(-5, 105)
    ax.set_xticks([0, 25, 50, 75, 100])
    
    ax.set_ylim(1.9, 3.2)
    ax.set_yticks([2.0, 2.4, 2.8, 3.2])
    
    # 设置刻度字体大小
    ax.tick_params(axis='both', which='major', labelsize=24)
    
    # 设置图例
    ax.legend(fontsize=24, loc='upper right')
    
    # 设置边框粗细
    for spine in ax.spines.values():
        spine.set_linewidth(2)

# 调整布局
plt.tight_layout()

# 保存图片
os.makedirs('figures', exist_ok=True)
output_file = 'figures/md_conver_gags_rg.tiff'
plt.savefig(output_file, dpi=300, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import os

matplotlib.rcParams['axes.unicode_minus'] = False

# 定义文件路径
rg_file = 'Nano20/b-rg.xvg'
sasa_file = 'Nano20/b-sasa.xvg'

def read_xvg_data(file_path, data_type='rg'):
    """读取xvg文件，返回时间和数值"""
    time_ps = []
    values = []
    
    with open(file_path, 'r') as f:
        for line in f:
            if line.startswith('#') or line.startswith('@'):
                continue
            data = line.strip().split()
            if len(data) >= 2:
                time_ps.append(float(data[0]))
                values.append(float(data[1]))
    
    return np.array(time_ps), np.array(values)

# 读取数据
time_ps_rg, rg_values = read_xvg_data(rg_file)
time_ps_sasa, sasa_values = read_xvg_data(sasa_file)

# 转换为纳秒
time_ns_rg = time_ps_rg / 1000.0
time_ns_sasa = time_ps_sasa / 1000.0

# 创建1×2子图
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 左图：Rg
ax1.plot(time_ns_rg, rg_values, color="#221F20", linewidth=1.5)
ax1.set_xlabel('Time (ns)', fontsize=28)
ax1.set_ylabel('Rg (nm)', fontsize=28)
ax1.tick_params(axis='both', which='major', labelsize=24)

# 右图：SASA
ax2.plot(time_ns_sasa, sasa_values, color="#221F20", linewidth=1.5)
ax2.set_xlabel('Time (ns)', fontsize=28)
ax2.set_ylabel('SASA (nm²)', fontsize=28)
ax2.tick_params(axis='both', which='major', labelsize=24)

for ax in (ax1, ax2):
    for spine in ax.spines.values():
        spine.set_linewidth(1.5)

plt.tight_layout()

# 保存图片
os.makedirs('figures', exist_ok=True)
plt.savefig('figures/nano20_rg_sasa.tiff', dpi=300, bbox_inches='tight')
plt.show()
```

# Compress

```python
import os
from PIL import Image
import glob

# 读取figures文件夹中所有tiff文件
for file_path in glob.glob('figures/*.tiff'):
    original_size = os.path.getsize(file_path) / (1024 * 1024)
    
    with Image.open(file_path) as img:
        # 方法1：使用libtiff压缩（效果更好，需要安装libtiff）
        output_path = file_path.replace('.tiff', '.tif')
        
        # 使用更高效的压缩方式
        img.save(output_path, 
                 format='TIFF', 
                 compression='tiff_lzw',  # LZW压缩
                 dpi=img.info.get('dpi', (300, 300)))  # 保持原dpi
    
    compressed_size = os.path.getsize(output_path) / (1024 * 1024)
    print(f"{os.path.basename(file_path)}: {original_size:.2f}MB → {compressed_size:.2f}MB (压缩率: {(1-compressed_size/original_size)*100:.1f}%)")

print("\n所有文件压缩完成！")
```
